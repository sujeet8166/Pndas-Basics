Q4. What is ‘DataFrame’ in pandas and how is it different from pandas.series? Explain with an example.

Ans:- A Series is a one-dimensional labeled array that can store data of any type.

A DataFrame is a two-dimensional labeled data structure with rows and columns. It can contain multiple Series objects

Difference Between Series and DataFrame

1. Dimension- Series is one-dimensional, DataFrame is two-dimensional.
2. Data Storage- Series stores data in a single column, DataFrame stores data in multiple columns.

3. Data Types- A Series contains one set of values, A DataFrame can contain different data types in different columns.

4. Structure- Series has only an index and values, DataFrame has rows, columns, and values.


Example-


```python

import pandas as pd

s = pd.Series([10, 20, 30, 40])

print(s)
```

DataFrame-

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Claire"],
    "Age": [25, 30, 27]
})

print(df)
```

Q5. What are some common functions you can use to manipulate data in a Pandas DataFrame? Can
you give an example of when you might use one of these functions?

Ans:- Some commonly used Pandas DataFrame functions are:

1. head() – Displays first rows.
2. tail() – Displays last rows.
3. info() – Shows information about the dataset.
4. describe() – Provides statistical summary.
5. drop() – Removes rows or columns.
6. fillna() – Fills missing values.
7. sort_values() – Sorts data.
8. groupby() – Groups similar records.

    Example:-

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Claire"],
    "Age": [25, 30, 27]
}

df = pd.DataFrame(data)

print(df.head())
```

Q6. Which of the following is mutable in nature Series, DataFrame, Panel?

Ans:- Series, DataFrame, and Panel are all mutable in nature.

This means their values can be modified after creation.

    Example:-

```python
import pandas as pd

s = pd.Series([10, 20, 30])

s[0] = 100

print(s)
```

Q7. Create a DataFrame using multiple Series. Explain with an example.

Ans:- A DataFrame can be created by combining multiple Series objects into a single table.

Example:- 

```python
import pandas as pd

name = pd.Series(["Alice", "Bob", "Claire"])
age = pd.Series([25, 30, 27])
gender = pd.Series(["Female", "Male", "Female"])

df = pd.DataFrame({
    "Name": name,
    "Age": age,
    "Gender": gender
})

print(df)
```
